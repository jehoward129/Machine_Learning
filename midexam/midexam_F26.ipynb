### Question 10 [10 points]

Separate the dataset into:

- Feature matrix `X`
- Target variable `y`

Then divide the data into training and testing sets using an appropriate train/test split.

```python
df['Classes  '] = df['Classes  '].str.strip()
df['Classes  '].unique()
df.dropna(inplace=True)
df.shape
```

```python
df['Classes_numeric'] = df['Classes  '].apply(lambda x: 1 if x == 'fire' else 0)
df['Classes_numeric'].unique()
```

```python
from sklearn.model_selection import train_test_split

# Create X and y

X = df[['day', 'month', 'year', 'Temperature', ' RH', ' Ws', 'Rain ', 'FFMC',
       'DMC', 'DC', 'ISI', 'BUI', 'FWI']]


y = df[['Classes_numeric']]

# Perform train/test split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=129)
```

### Question 11 [10 points]

Train a k-NN classifier using: k= 1..5
Report:

Accuracy

Confusion matrix

Train the model using the training data and generate predictions for the test data.

```python
# Import your model
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# Create the model
knn = KNeighborsClassifier(n_neighbors=5)
# Train the model
knn.fit(X_train, np.ravel(y_train))

# Generate predictions
y_pred = knn.predict(X_test)
acc = knn.score(X_test, np.ravel(y_test))
cm = confusion_matrix(y_test, y_pred)

print(acc)
print(cm)
```

### Question 12 [10 points]

Train a Logistic Regression classifier using standardized features.

Train the model using the training data and generate predictions for the test data.


```python
# Import your model
from sklearn.linear_model import LogisticRegression

# Create the model
log_model = LogisticRegression()
# Train the model
log_model.fit(X_train, np.ravel(y_train))
# Generate predictions
y_pred = log_model.predict(X_test)
print(log_model.score(X_test, np.ravel(y_test)))
print(confusion_matrix(y_test, y_pred))
```

### Question 13 [10 points]

Train a Gaussian Naive Bayes classifier.

### Question 15 [10 points]

Use FWI (Fire Weather Index) as the response variable. Remove Classes from the predictors, fit a Linear Regression model, and evaluate its performance on the test set using:

- MAE

- RMSE

- R²

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Your code here
X = df[['FWI']]
y = df[['Classes_numeric']]

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size= 0.33, random_state=129)

#Create and fit model
lin_model = LinearRegression()
lin_model.fit(X_train,y_train)
y_pred = lin_model.predict(X_test)
mae_score = mean_absolute_error(y_pred, y_test)
mse_score = mean_squared_error(y_pred, y_test)
r2 = r2_score(y_pred, y_test)

print(mae_score)
print(mse_score)
print(r2)
```
